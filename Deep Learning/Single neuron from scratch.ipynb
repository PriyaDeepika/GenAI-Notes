```python
import torch

x = torch.tensor([5.0, 8.0])

w = torch.tensor([0.6, 0.2], requires_grad = True)
b = torch.tensor(1.0, requires_grad = True)

y = (x*w).sum() + b
print("y: ", y)

target = torch.tensor(8.0)

loss = (y-target)**2
loss.backward()

print("loss: ", loss)
print("loss requires gradient: ", loss.requires_grad)
print("w: ", w.grad)
print("b: ", b.grad)
```

```python
#Using loop

x = torch.tensor([5.0, 8.0])

w = torch.tensor([0.6, 0.2], requires_grad = True)
b = torch.tensor(1.0, requires_grad = True)

target = torch.tensor(8.0)

lr = 0.01

for epoch in range(20):
    # Forward pass
    y = (x*w).sum() + b
    loss = (y - target)**2

    # Backward pass
    loss.backward()

    # Update parameters
    with torch.no_grad():
        w -= lr*w.grad
        b -= lr*b.grad

    # Clear gradients
    w.grad.zero_()
    b.grad.zero_()

    print(
        f"Epoch {epoch+1: 2d} | "
        f"Prediction = {y.item():.4f} | "
        f"Loss = {loss.item():.4f}"
    )
```

```python
X = torch.tensor([
    [1.0, 2.0],
    [2.0, 3.0],
    [3.0, 4.0],
    [4.0, 5.0]
])

y_true = torch.tensor([
    3.0,
    5.0,
    7.0,
    9.0
])

print(X.shape)
print(y_true.shape)
```

```python
w = torch.tensor([0.3, -0.5], requires_grad = True)
b = torch.tensor(2.0, requires_grad = True)

lr = 0.01

for epoch in range(1000):
    y_pred = X@w + b
    loss = ((y_pred - y_true) ** 2).mean()

    loss.backward()
    with torch.no_grad():
        w -= lr*w.grad
        b -= lr*b.grad

    w.grad.zero_()
    b.grad.zero_()

    print(
        f"Epoch {epoch+1: 2d} | "
        f"Prediction = {y_pred} | "
        f"Loss = {loss.item():.4f}"
    )
```
